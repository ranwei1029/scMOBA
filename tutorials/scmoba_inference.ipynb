# Original scMOBA: human M1 broad-class inference (staging)


Run from the repository root in the original-model inference environment. Supply the matched original-model assets and an authorized M1 dataset tokenized with the original scMOBA workflow.

```python
from pathlib import Path
import os, subprocess, sys

assets = {'dataset': 'data/m1-tokenized', 'checkpoint': 'models/original/model.bin',
          'tokenizer': 'models/original/tokenizer', 'geneformer': 'models/original/geneformer',
          'llm': 'models/original/llm'}
assets = {k: os.getenv(f'SCMOBA_{k.upper()}', v) for k, v in assets.items()}
assert all(Path(v).exists() for v in assets.values()), assets
output = Path(os.getenv('SCMOBA_OUTPUT', 'm1_class_predictions.csv'))
cmd = [sys.executable, '-m', 'src.test.infer_m1', '--output', str(output), '--limit', '8']
for key, value in assets.items():
    cmd.extend(['--' + key, value])
subprocess.run(cmd, check=True)
```
